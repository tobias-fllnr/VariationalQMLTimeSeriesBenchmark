```python
import reservoirpy.datasets
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

plt.style.use("seaborn-v0_8-deep")
plt.rcParams.update(
    {
        "text.usetex": True,  # Use LaTeX for text rendering
        "font.family": "serif",  # Use serif font
        "font.serif": ["Computer Modern Roman"],  # Use default LaTeX font
    }
)
```

Generate Lorenz Data

```python
time_steps = 10000
data = reservoirpy.datasets.lorenz(time_steps)
df = pd.DataFrame({"x_coords": data[:, 0], "y_coords": data[:, 1], "z_coords": data[:, 2]})
df_data = df.loc[500:1499, :]  # leave out first 500 time steps
df_data.to_csv(f"lorenz_1000.csv", index=False)
```

Plot Lorenz Data

```python
data_lorenz = np.array(df_data)

x = data_lorenz[:, 0]
y = data_lorenz[:, 1]
z = data_lorenz[:, 2]

fig, axes = plt.subplots(3, 1, figsize=(1.8, 1.8), sharex=True)

# Plot x(t)
axes[0].plot(x[: 19 * 15], color='black', linewidth=1.3)

# Plot y(t)
axes[1].plot(y[: 19 * 15], color='black', linewidth=1.3)

# Plot z(t)
axes[2].plot(z[: 19 * 15], color='black', linewidth=1.3)

fig.text(0.008, 0.85, "$x(t_{step})$", va="center", rotation="vertical", fontsize=10)
fig.text(0.008, 0.58, "$y(t_{step})$", va="center", rotation="vertical", fontsize=10)
fig.text(0.008, 0.31, "$z(t_{step})$", va="center", rotation="vertical", fontsize=10)
fig.text(0.55, 0.03, "$t_{step}$", ha="center", fontsize=10)
# Adjust spacing between subplots
plt.tight_layout()
fig.subplots_adjust(left=0.28, right=0.98, top=0.975, bottom=0.18, wspace=0.25, hspace=0.15)
plt.savefig("lorenz_timeseries.pdf")
```
