```python
import reservoirpy.datasets
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

plt.style.use("seaborn-v0_8-deep")
plt.rcParams.update(
    {
        "text.usetex": True,  # Use LaTeX for text rendering
        "font.family": "serif",  # Use serif font
        "font.serif": ["Computer Modern Roman"],  # Use default LaTeX font
    }
)
```

Generate Henon Data

```python
time_steps = 10000
data = reservoirpy.datasets.henon_map(time_steps)
df = pd.DataFrame({"x_coords": data[:, 0], "y_coords": data[:, 1]})
df_data = df.loc[:999, :]
df_data.to_csv(f"henon_1000.csv", index=False)
```

Plot Henon Data

```python
data_henon = np.array(df_data)

x = data_henon[:, 0]
y = data_henon[:, 1]

fig, axes = plt.subplots(2, 1, figsize=(1.8, 1.8), sharex=True)

# Plot x(t)
axes[0].plot(x[: 15 * 4])
# axes[0].set_ylabel('$x(t_{step})$')

# Plot y(t)
axes[1].plot(y[: 15 * 4])
# axes[1].set_ylabel('$y(t_{step})$', fontsize=10)

# Set common x-axis label
# axes[1].set_xlabel('$t_{step}$', fontsize=10)
# fig.suptitle('Hénon Map Time Series Data')
fig.text(0.008, 0.78, "$x(t_{step})$", va="center", rotation="vertical", fontsize=10)
fig.text(0.008, 0.36, "$y(t_{step})$", va="center", rotation="vertical", fontsize=10)
fig.text(0.62, 0.03, "$t_{step}$", ha="center", fontsize=10)
# Adjust spacing between subplots
plt.tight_layout()
fig.subplots_adjust(left=0.35, right=0.98, top=0.975, bottom=0.165, wspace=0.25, hspace=0.15)
plt.savefig("henon_timeseries.pdf")
```
